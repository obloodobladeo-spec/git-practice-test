### 내가 짠 코드와 강사님 코드 비교
01. 내가 짠 코드

```python
class Usererror(Exception):
    pass

class Account:
    def __init__(self, account_id, name, balance):
        self.account_id = account_id
        self.name = name
        self.balance = balance

    def in_money(self, money):
        self.balance = self.balance + money
        return self.balance

    def out_money(self, money):
        self.balance = self.balance - money
        return self.balance

    def show_info(self):
        print("계좌 ID : {}".format(self.account_id))
        print("이름 : {}".format(self.name))
        print("잔액 : {}".format(self.balance))



acc_lst = []

while(True):
    print("-----menu-----")
    print("1. 계좌개설")
    print("2. 입금")
    print("3. 출금")
    print("4. 계좌정보 보기")
    print("5. 프로그램 종료")
    try:
        anwer = int((input("선택(1~5까지의 숫자만 입력) : ")))
        if anwer == 1:
            id = int(input("계좌ID(숫자 4자리) :"))
            if len(str(id)) != 4:
                raise Usererror
            name = input("성함을 입력해주세요 : ")
            acc_lst.append(Account(id, name, 0))
            print("[계좌개설]")
            print(acc_lst)

        elif anwer == 2:
            id = int(input("계좌ID(숫자 4자리) :"))
            if len(str(id)) != 4:
                raise Usererror
            money = int(input("입금할 금액 : "))
            acc_num = []
            for acc in acc_lst:
                acc_num.append(acc.account_id)
                if acc.account_id == id:
                    balance = acc.in_money(money)
                    print("입금액 : {}".format(money))
                    print("계좌잔금 : {}".format(balance))
                    print("입금완료")
            if id not in acc_num:
                print("유효하지 않은 ID입니다.")

        elif anwer == 3:
            id = int(input("계좌ID(숫자 4자리) :"))
            if len(id) != 4:
                raise Usererror
            money = int(input("입금할 금액 : "))
            acc_num = []
            for acc in acc_lst:
                acc_num.append(acc.account_id)
                if acc.account_id == id:
                    balance = acc.out_money(money)
                    print("입금액 : {}".format(money))
                    print("계좌잔금 : {}".format(balance))
                    print("출금완료")
            if id not in acc_num:
                print("유효하지 않은 ID입니다.")

        elif anwer == 4:
            id = int(input("계좌ID(숫자 4자리) :"))
            if len(id) != 4:
                raise Usererror
            for acc in acc_lst:
                if acc.account_id == id:
                    acc.show_info()

        elif anwer == 5:
            print("저희 은행은 이용해주셔서 감사합니다.")
            break

        else:
            raise Usererror

    except ValueError:
        print("에러 : 숫자 값만 입력해주세요.")
        print(input("Enter키를 누르세요."))

    except Usererror:
        print("에러 : 잘못된 값을 입력하였습니다.")
        print(input("Enter키를 누르세요."))

# - 계좌번호 당 구별되는 객체를 만들기 위하여 Class를 부여함
# 어려운점 : 각 클래스의 id 만들 구별하는 방법을 몰라서 어려움을 겪음
ㄹ
```

```python
class Account:
    def __init__(self, acc_id, balance, name):
        self._acc_id = acc_id
        self._balance = balance
        self._name = name

    def get_acc_id(self):
        return self._acc_id

    def deposit(self, money):
        self._balance += money

    def withdraw(self, money):
        if self._balance < money:
            return 0
        self._balance -= money
        return money

    def show_info(self):
        print(f"계좌ID: {self._acc_id}")
        print(f"이름: {self._name}")
        print(f"잔액: {self._balance}\n")

acc_arr = []  

def show_menu():
    print("-----Menu-----")
    print("1. 계좌개설")
    print("2. 입금")
    print("3. 출금")
    print("4. 계좌번호 전체 출력")
    print("5. 프로그램 종료")

def make_account():
    print("[계좌개설]")
    try:
        acc_id = int(input("계좌ID:(숫자로 입력) "))
        name = input("이름: ")
        balance = int(input("입금액: "))
        print()
    except ValueError:
        print("\n입력 형식이 올바르지 않습니다.\n")
        return

    acc_arr.append(Account(acc_id, balance, name))

def deposit_money():
    print("[입  금]")
    try:
        acc_id = int(input("계좌ID: "))
        money = int(input("입금액: "))
    except ValueError:
        print("\n입력 형식이 올바르지 않습니다.\n")
        return

    for acc in acc_arr:
        if acc.get_acc_id() == acc_id:
            acc.deposit(money)
            print("입금완료\n")
            return
    print("유효하지 않은 ID 입니다.\n")

def withdraw_money():
    print("[출  금]")
    try:
        acc_id = int(input("계좌ID: "))
        money = int(input("출금액: "))
    except ValueError:
        print("\n입력 형식이 올바르지 않습니다.\n")
        return

    for acc in acc_arr:
        if acc.get_acc_id() == acc_id:
            if acc.withdraw(money) == 0:
                print("잔액부족\n")
                return
            print("출금완료\n")
            return
    print("유효하지 않은 ID 입니다\n")

def show_all_acc_info():
    for acc in acc_arr:
        acc.show_info()

def main():
    MAKE, DEPOSIT, WITHDRAW, INQUIRE, EXIT = 1, 2, 3, 4, 5
    while True:
        show_menu()
        try:
            choice = int(input("선택(1~5까지의 숫자만 입력) : "))
        except ValueError:
            print("\n잘못된 선택\n")
            continue
        print()

        if choice == MAKE:
            make_account()
        elif choice == DEPOSIT:
            deposit_money()
        elif choice == WITHDRAW:
            withdraw_money()
        elif choice == INQUIRE:
            show_all_acc_info()
        elif choice == EXIT:
            break
        else:
            print("잘못된 선택\n")


# if __name__ == "__main__":
#     main()

# 각 기능을 따로 함수를 만들어서 정리를 함
# 강사님의 성향상 c언어 기반의 코드가 작성 되었을 확률이 있음(class와 함수, main을 따로 둠)
```
