## **생성자 __init__**
각 객체가 기본적으로 가지게 되는 초기화작업

```python
class Student:
    def __init__(self, name, korean, math, english, science):
        self.name = name
        self.korean = korean
        self.math = math
        self.english = english
        self.science = science

# 이때, self 는 객체 이름으로 바뀐다.
s = Student("윤인성", 87, 98, 88, 95)
# 예
# class Student:
#     def __init__(s1, name, korean, math, english, science):
#         s1.name = name
#         s1.korean = korean
#         s1.math = math
#         s1.english = english
#         s1.science = science
```

```python
class Student:
    def __init__(self, name, korean, math, english, science):
        self.name = name
        self.korean = korean
        self.math = math
        self.english = english
        self.science = science

    def get_sum(self):
        return self.korean + self.math +\
            self.english + self.science

    def get_average(self):
        return self.get_sum() / 4

    def to_string(self):
        return "{}\t{}\t{}".format(\
            self.name,\
            self.get_sum(),\
            self.get_average())

```

```python
students = [
    Student("윤인성", 87, 98, 88, 95),
    Student("연하진", 92, 98, 96, 98),
    Student("구지연", 76, 96, 94, 90),
    Student("나선주", 98, 92, 96, 92),
    Student("윤아린", 95, 98, 98, 98),
    Student("윤명월", 64, 88, 92, 92)
]

print("이름", "총점", "평균", sep="\t")
for student in students:
    print(student.to_string())
```

# 상속
부모클래스와 자식클래스

```python
class Father:
    def run(self):
        print("달린다!")

# Son 클래스는 Father 클래스를 상속한다!
class Son(Father):
    # Father 메소드가 포함되어 있음
    # def run(self):
    #     print("달린다!")
    def jump(self):
        print("점프한다!")
```

```python
s = Son()
s.run() # father 메소드
s.jump() # son 메소드
```

## 다중 상속
둘 이상의 클래스를 상속 받을 수 있음

```python
class Father:
    def run(self):
        print("달린다!")

class Mother:
    def swim(self):
        print("헤엄친다!")

class Son(Father, Mother):
    def jump(self):
        print("점프한다!")
```

```python
s = Son()
s.run()
s.swim()
s.jump()
```

## 메소드 오버라이딩
- 부모 클래스가 갖는 메소드와 동일한 이름의 메소드를 자식 클래스가 정의하는 경우  
- 이 경우 부모 클래스의 메소드는 자식 클래스에서 호출할 수 없는 상태가 됩

```python
class Father:
    def run(self):
        print("달린다!")

# 메소드 오버라이딩 : 똑같은 제목을 가진 함수를 제정의
class Son(Father):
    def run(self):
        print("나만의 스타일로 펀-런한다!")
```

## super
부모 클래스에 접근하고자 할 때 호출하는 함수

```python
# 위 예제처럼 메소드 오버라이딩 작업을 하고 나면 부모 클래스의 메소드를 호출할 수 없는 상태가 되지만, 
# 완전히 불가능한 것은 또 아닙니다

class Father:
    def run(self):
        print("달린다!")

class Son(Father):
    def run(self):
        # 부모 클래스의 run() 함수를 먼저 호출하자, 부모를 호출하는 형태임으로 함수와 같이 ()를 써주어야 한다.
        super().run()

        print("나만의 스타일로 펀-런한다!")

s = Son()
s.run()
```
